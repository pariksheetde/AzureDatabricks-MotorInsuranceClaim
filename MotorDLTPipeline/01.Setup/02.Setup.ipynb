#### CREATE VOLUME

```sql
USE CATALOG motor;
USE SCHEMA landing;
CREATE EXTERNAL VOLUME IF NOT EXISTS operational
LOCATION 'abfss://motor@motorinsurancedevadls.dfs.core.windows.net/landing/operational'
COMMENT 'CREATE VOLUME opearation_data';
```

```sql
USE CATALOG motor;
USE SCHEMA landing;
CREATE EXTERNAL VOLUME IF NOT EXISTS orders
LOCATION 'abfss://motor@motorinsurancedevadls.dfs.core.windows.net/landing/order_status'
COMMENT 'CREATE VOLUME orders';
```
