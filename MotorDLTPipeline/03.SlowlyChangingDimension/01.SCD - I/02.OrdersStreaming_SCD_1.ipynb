#### IMPLEMENT STREAMING
- SOURCE `MOTOR.SILVER.ORDER_STATUS_RAW_SCD_1`
- TARGET `MOTOR.SILVER.ORDER_STATUS_STREAMING_SCD_1`

```python
import pyspark.sql.functions as F

(spark.readStream.option('readChangeFeed', 'true')
                .table('MOTOR.SILVER.ORDER_STATUS_RAW_SCD_1')
                .filter(F.col('_change_type') != 'update_preimage')
                .writeStream.outputMode('append')
                .option('checkpointLocation', '/checkpoint10/orders_status')
                .table('MOTOR.SILVER.ORDER_STATUS_STREAMING_SCD_1')
    )            
```
